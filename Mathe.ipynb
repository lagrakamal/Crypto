## Verknüpfung

Sei *M* eine Menge. Eine Verknüpfung auf *M* ist eine Abbildung

*   : **M × M -> M**

die je zwei Elementen a, b elemente aus der menge M ein Element c aus der menge M zuordnet:

**a • b  =  c** 

Wenn die Menge M endlich ist, lassen sich die Ergebniswerte der Verknüpfung in Form einer Verknüpfungstafel angeben.

```python
def vTafel(n,v):
    head   = f' {v} |'
    line = '___|' + n*'___'
    for a in range(n):
        head = head + format(a,'3d')
    print(head)
    print(line)
    
    for a in range(1,n):
        row = format(a,'2d')+' |'
        for k in range(n):
            if (v == '+'):
                row = row + format((a+k)%n,'3d')
            elif (v == '*'):
                row = row + format((a*k)%n,'3d')
            elif (v == '^'):
                row = row + format((a**k)%n,'3d')
        print(row)
```

```python
vTafel(9,"+")
```

```python
vTafel(9,"*")
```

```python
vTafel(9,"^")
```

### Eigenschaften von Verknüpfungen

* Kommutativ

Sei M eine Menge mit einer Verknüpfung  •. 
Die Verknüpfung  •  heißt kommutativ, wenn für alle Elemente a, b Element aus der Menge M gilt:

**a • b  =  b • a**

Nicht Kommutative verknüpfungsbeispiele: 2³ = 8, aber 3² = 9 || 2 - 3 = -1, aber 3 - 2 = 1

* Assoziative

Sei M eine Menge mit einer Verknüpfung  • . Die Verknüpfung  •  heißt assoziativ, wenn für alle Elemente a, b, c Elemente aus der Menge M gilt

**(a • b) • c  =  a • (b • c)**

Nicht Assoziative verknüpfungsbeispiele: (2^3)^2 = 8^2 aber 2^(3^2) = 2^9 || (2 * 5) - 4 = 6 aber  2 * (5 - 4) = 2

### Halbgruppe

Definition:  Eine Menge mit einer assoziativen Verknüpfung ist eine Halbgruppe.

### Monoid

Eine Halbgruppe (M,  • ) ist ein Monoid, wenn sie ein neutrales Element hat. Ein Element e aus der menge M heißt neutrales Element, wenn für alle a aus der menge M gilt:

**e • a  =  a • e  =  a**

Das neutrale Element oder Identitäts Element bezeichnet, je nach dem, ob die Verknüpfung als Addition oder als Multiplikation verstanden wird.

Beispiele: 

a * 1 = a und 1 * a = a

a + 0 = a und 0 + a = a

### Gruppe

Ein Monoid (M,  • , e) ist eine Gruppe, wenn es zu jedem Element a aus der menge M ein inverses Element in M gibt. Ein Element a' aus der menge M heißt invers zu a aus der menge M, wenn:

**a • a'  =  e**

d.h wenn die Verknüpfung von a und a' das neutrale Element e ergibt

Die Menge der Z (ganzen Zahlen) mit der Addition (Z, +, 0) ist eine Gruppe. Zu jeder Zahl a ist -a das inverse Element.

Die Menge der Z (ganzen Zahlen) mit der Multiplikation (Z, *, 1) ist eine Gruppe. Zu jeder Zahl a ist a^-1 das inverse Element.

### Abelsche Gruppe

Eine Gruppe (M, • , e) heißt abelsche Gruppe, wenn die Verknüpfung • kommutativ ist, d.h. wenn für alle Elemente a, b aus der menge M gilt:

**a • b  =  b • a**

Die additive Gruppe modulo n (Z, +n, 0) ist eine abelsche Gruppe. Die Menge ganze Zahlen besteht aus den ganzen Zahlen 0, ..., n-1. Die Verknüpfung +n bezeichnet die Addition modulo n.

Die multiplikative Gruppe modulo n (Zn*, ·n, 1) ist eine abelsche Gruppe. Die Menge Zn* besteht aus allen Elementen von Zn, die teilerfremd zu n sind. Die Verknüpfung ·n bezeichnet die Multiplikation modulo n.

#### Zusammenfassung

Eine Gruppe ist zunächst eine Menge M, auf der eine Verknüpfung  •  definiert ist. Eine Verknüpfung ist eine Abbildung von M × M in M. 

Es müssen darüber hinaus drei Eigenschaften erfüllt sein, diese heißen die Gruppenaxiome:

**die Verknüpfung ist assoziativ,**

**es gibt ein neutrales Element,**

**jedes Element hat ein inverses Element.**

Bei einer abelschen Gruppe gilt zusätzlich:

**die Verknüpfung ist kommutativ.**

### Generator

Sei (G, • , e) eine endliche Gruppe und a ein Element von G.

Die von a erzeugte Untergruppe ist

< a > = { a^k | k  Element von N}

< a> = { a^k | k  Element von N}

Beispiel:  Die Menge Z6 = {0, 1, 2, 3, 4, 5} bildet mit der Operator ** , folgende Generatoren : {1, 5}

```python
def generators(p,v,e):
    s = set(range(e, p))
    res = []
    print(f"check for generators in Z{p}{v}")
    for a in s:
        g = set()
        for x in s:
            if(v == "*"):
                g.add((a*x) % p)
            elif(v == "+"):
                g.add((a+x) % p)
            elif(v == "**"):
                g.add((a**x) % p)
        print(f"<{a}> = {g}")
        if g == s:
            res.append(a)
    return set(res)
```

### Ring

Sei (M, +, ·) eine Menge mit zwei Verknüpfungen. M ist ein Ring, wenn folgende Bedingungen erfüllt sind:

**(M, +, 0) ist eine abelsche Gruppe,**

**(M, ·) ist eine Halbgruppe,**

es gilt das Distributivgesetz, d.h. für alle a, b, c Element aus der menge M gilt:

**a·(b + c)  =  (a·b) + (a·c) sowie (a + b)·c  =  (a·c) + (b·c)**

### Die Multiplikative Gruppe Zn*

Sei n Element aus der Menge N. Die Menge Zn* besteht aus allen Elementen von Zn, die teilerfremd zu n sind. Mit phi(n) wird die Anzahl der Elemente von Zn* dargestellt.

**Zn\*  =  { k Element aus Z  |  ggt(k, n) = 1 }**

**Phi(n)  =  |Zn\*|**

Beispiel:

Z10*  =  { 1, 3, 7, 9 }\
Z15*  =  { 1, 2, 4, 7, 8, 11, 13, 14 }

phi(10) = 4 \
phi(15) = 8

Wenn n eine Primzahl ist, gilt **phi(n)  =  n-1**

Z7*   =  { 1, 2, 3, 4, 5, 6 }\
phi(7) = (7-1) = 6

Wenn n das Produkt zweier verschiedener Primzahlen p, q ist | n = p·q, so gilt **phi(n)  =  (p-1)(q-1)**

phi(21) = (3– 1)×(7– 1) = 2×6 = 12

### Satz von Euler

Wenn n und a teilerfremde positive ganze Zahlen
sind (d.H. ggT(a, n) = 1), dann gilt

a^phi(n) mod n ≡ 1

Beispiele:

a = 3, n = 10 , phi(10) = 4\
3^4 = 81 ≡ 1 mod 10

a = 2, n = 11, phi(11) = 10\
2^10 = 1024 ≡ 1 mod 11

pyBeispiele:

```python
def factors(n):
    i = 2
    factors = []
    while i * i <= n:
        if n % i:
            i += 1
        else:
            n //= i
            factors.append(i)
    if n > 1:
        factors.append(n)
    return factors
```

```python
def ggt(a,b):
    if not b:
        return a
    return ggt(b,a%b)
```

```python
def ggt_phi(n):
    result = 1
    for i in range(2,n):
        if ggt(i,n) == 1:
            result += 1
    return result
```

```python
def phi(n):
    result = n
    i = 2
    while i*i <= n:
        if n % i == 0:
            result -= result//i
            while n % i == 0:
                n //= i
        i += 1
    if n != 1:
        result -= result//n
    return result
```

### Inverses Element modulo n

Berechnung mit dem erweiterten euklidischen Algorithmus

Die multiplikative Gruppe  Zn* besteht aus den Elementen aus der Menge Z, die teilerfremd zu n sind. Für jedes a von Zn* gilt dann **ggt(a, n) = 1**

und die 1 lässt sich nach dem Lemma von Bézout als ganzzahlige Linearkombination von a und b darstellen:

1 = u·a + v·n wobei u, v elemente aus der Menge Z sind

1 ≡ u·a + v·n ≡  u·a  mod n

Multiplikation mit a^-1 ergibt

a^-1 ≡ u mod n

pyBeispiel:

```python
def erggt(a, b):
    if (b==0):
        return a, 1, 0
    else:
        g, u, v = erggt(b, a%b)
        q=a//b
        print(f"{g} = {v} * {a} + ({u} - {q} * {v} ) * {b}") 
        return g, v, u-q*v
```

```python
def modinvers(a, n):
    g, u, v=erggt(a, n)
    return u%n
```
