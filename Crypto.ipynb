## Modulo_Arithmetik:

* Addition

```python
def modAdd(x,y,p):
    number = add(x,y)
    q = number//p
    rest = sub(number,mult(q,p))
    print("a = x  y = ",number,"| a mod p => q * p + (r) |","q = ",q,"| p = ",p,"| r = ", rest,"\n")
    print(f"Results in Z{p} : {number} ≡ {rest} Mod {p}")
    return rest
```

```python
modAdd(6,11,6)
```

* Substitition

```python
def modSub(x,y,p):
    number = sub(x,y)
    q = number//p
    rest = sub(number,mult(q,p))
    print("a = x - y = ",number,"| a mod p => q * p + (r) |","q = ",q,"| p = ",p,"| r = ", rest,"\n")
    print((f"Results in Z{p} : {number} ≡ {rest} Mod {p}"))
    return rest
```

```python
modSub(1024,3,6)
```

 ## Modulus p, zyklische multiplikative Gruppe mit generator g

```python
def generateSomeset(p, n):
    return set(i for i in range(p) if i % n == 0)
```

```python
generateSomeset(100,9)
```

```python
def generateprimeset(n=100):
    return set(x for x in range(2, n) if not any(x % y == 0 for y in range(2, x)))
```

```python
generateprimeset(50)
```

* potenzieren 

```python
def powGenerators(p):
    s = set(range(1, p))
    res = []
    for a in s:
        g = set()
        for x in s:
            g.add(pow(a,x) % p)
        if g == s:
            res.append(a)
    return set(res)
```

```python
def printPowGenerators(p):
    for i in range(p):
        gens = powGenerators(i)
        if gens:
            print(f"Z{i} has following generators : {gens}")
```

```python
printPowGenerators(10)
```

* Multiplizieren

```python
def multGenerators(p):
    s = set(range(1, p))
    res = []
    for a in s:
        g = set()
        for x in s:
            g.add(mult(a,x) % p)
        if g == s:
            res.append(a)
    return set(res)
```

```python
def printMultGenerators(p):
    for i in range(p):
        gens = multGenerators(i)
        if gens:
            print(f"Z{i} has following generators : {gens}")
```

```python
printMultGenerators(6)
```

### ggt berechnen | Euklidischer Algorithmus

##### Lemma von Bézout : ggt(a, b)  = ggt(b, a mod b) 

* g  =  u·b + v·(a mod b)

##### a-q·b = a mod b && q = a // b

* g  =  u·b + v·(a-q·b)

##### v·(a-q·b) = v·a – v·q·b (ausklammern)

* g  =  u·b + v·a – v·q·b

##### u·b – v·q·b = (u-q·v)·b (einklammern)

* g  =  v·a + (u-q·v)·b



```python
def ggt(a, b):
    if (b==0):
        return a, 1, 0
    else:
        g, u, v = ggt(b, a%b)
        q=a//b
        #print(g,u,v)
        print(f"Results {g} = {v} * {a} + ({u} - {q} * {v} ) * {b}") 
        return g, v, u-q*v
```

```python
ggt(7,29)
```

```python
def factors(n):
    i = 2
    factors = []
    while (i * i <= n):
        if n % i:
            i += 1
        else:
            n = n//i
            factors.append(i)
    if n > 1:
        factors.append(n)
    return factors
```
